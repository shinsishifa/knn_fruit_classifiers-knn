```python
weight= [150,170,160,130,140,120]
sweet_score =[7,8,7,4,5,3]
fruits = ["apple","apple","apple","orange","orange","orange"]
```

```python
import pandas as pd
df=pd.DataFrame({"weight":weight ,"sweet_score":sweet_score,"fruits":fruits})
df
```

```python
features=df.iloc[:,0:2].values
features
```

```python
targets=df.iloc[:,-1].values
targets
```

```python
from sklearn.neighbors import KNeighborsClassifier
model=KNeighborsClassifier(n_neighbors=3)
model.fit(features,targets)
```

```python
model.predict([[155,7]])
```
